```python
import pandas as pd
import plotly.graph_objects as go

# 1. Create simulated historical and forecast data for your Class 9 business
data = {
    'Year': [2022, 2023, 2024, 2025, 2026, 2027, 2028],
    'Metric_Value': [120, 195, 310, 480, 650, 890, 1200],  # e.g., active deployments, units sold, or revenue in ZAR (thousands)
    'Type': ['Historical', 'Historical', 'Historical', 'Historical', 'Forecast', 'Forecast', 'Forecast'],
    'Upper_Bound': [None, None, None, None, 720, 980, 1350],  # Confidence interval for forecast
    'Lower_Bound': [None, None, None, None, 580, 800, 1050]   # Confidence interval for forecast
}

df = pd.DataFrame(data)

# Split into historical and forecast subsets for separate styling
df_hist = df[df['Type'] == 'Historical']
df_fcst = df[df['Type'] == 'Forecast']

# 2. Initialize the Plotly figure
fig = go.Figure()

# Add Historical Data Trace
fig.add_trace(go.Scatter(
    x=df_hist['Year'],
    y=df_hist['Metric_Value'],
    mode='lines+markers',
    name='Historical Performance',
    line=dict(color='#1f77b4', width=3),
    marker=dict(size=8)
))

# Add Forecast Data Trace
fig.add_trace(go.Scatter(
    x=df_fcst['Year'],
    y=df_fcst['Metric_Value'],
    mode='lines+markers',
    name='Projected Forecast',
    line=dict(color='#ff7f0e', width=3, dash='dash'),
    marker=dict(size=8)
))

# Add Forecast Confidence Interval (Upper & Lower bounds)
fig.add_trace(go.Scatter(
    x=pd.concat([df_fcst['Year'], df_fcst['Year'][::-1]]),
    y=pd.concat([df_fcst['Upper_Bound'], df_fcst['Lower_Bound'][::-1]]),
    fill='tozeroy',
    fillcolor='rgba(255, 127, 14, 0.15)',
    line=dict(color='rgba(255,255,255,0)'),
    showlegend=False,
    name='Confidence Interval'
))

# 3. Update layout for interactivity and clean presentation
fig.update_layout(
    title='<b>Nice Class 9 Trademark Growth & Forecast Model</b><br><i>Performance Tracking & Projections</i>',
    xaxis_title='Year',
    yaxis_title='Performance Metric (Units / Revenue Index)',
    template='plotly_white',
    hovermode='x unified',
    legend=dict(x=0.02, y=0.98),
    xaxis=dict(
        rangeslider=dict(visible=True),
        type='linear'
    )
)

# 4. Display the interactive chart
fig.show()

# Optional: To save the chart as an interactive HTML file to open in any browser
# fig.write_html("class9_trademark_forecast.html")
```
