```python
import math
import matplotlib.pyplot as plt
import networkx as nx

# Initialize directed graph
G = nx.DiGraph()

# Central Hub
root = "AI-to-Lean\nMath Pipeline"
G.add_node(root)

# Categories and Sub-nodes
categories = {
    "1. Input & Generation": [
        "Natural Language Prompt",
        "LLM Reasoning / Search",
        "Informal Proof Output",
    ],
    "2. Autoformalization": [
        "Translate to Lean Syntax",
        "Type Inference & Mapping",
        "Silent Error Insertion\n(The Paper's Risk)",
    ],
    "3. Lean Execution": [
        "Lean 4 Kernel Check",
        "mathlib Integration",
        "Proof State Validation",
    ],
    "4. Measurement & Feedback": [
        "Syntactic Success\n(Lean Accepts)",
        "Semantic Equivalence Check\n(Did it alter the claim?)",
        "RL Reward / Optimization",
    ],
}

# Layout Calculations
pos = {root: (0, 0)}
cat_keys = list(categories.keys())
n_cats = len(cat_keys)
radius_cat = 2.5
radius_node = 4.5

for i, cat in enumerate(cat_keys):
  angle = 2 * math.pi * i / n_cats
  cx = radius_cat * math.cos(angle)
  cy = radius_cat * math.sin(angle)
  pos[cat] = (cx, cy)
  G.add_edge(root, cat)

  nodes = categories[cat]
  n_nodes = len(nodes)
  for j, node in enumerate(nodes):
    node_angle = angle + (j - (n_nodes - 1) / 2) * 0.35
    nx_pos = radius_node * math.cos(node_angle)
    ny_pos = radius_node * math.sin(node_angle)
    pos[node] = (nx_pos, ny_pos)
    G.add_edge(cat, node)

# Plotting
plt.figure(figsize=(10, 10), dpi=300)
nx.draw_networkx_nodes(
    G, pos, nodelist=[root], node_size=3000, node_color="#1f77b4", alpha=0.9
)
nx.draw_networkx_nodes(
    G, pos, nodelist=cat_keys, node_size=2200, node_color="#ff7f0e", alpha=0.85
)

leaf_nodes = [n for n in G.nodes() if n != root and n not in cat_keys]
nx.draw_networkx_nodes(
    G, pos, nodelist=leaf_nodes, node_size=1600, node_color="#2ca02c", alpha=0.8
)

nx.draw_networkx_edges(G, pos, width=2, alpha=0.6, edge_color="gray")
nx.draw_networkx_labels(
    G, pos, font_size=9, font_family="sans-serif", font_weight="bold"
)

plt.title(
    "AI-to-Lean Mathematical Verification & Simulation Pipeline",
    fontsize=14,
    fontweight="bold",
    pad=20,
)
plt.axis("off")
plt.tight_layout()
plt.savefig("lean_ai_mindmap.png", dpi=300, bbox_inches="tight")
plt.show()
```
