```python
import math
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Pendulum parameters
origin_x, origin_y = 300, 150
length = 250
angle = math.pi / 3  # Initial angle (60 degrees)
angular_velocity = 0.0
gravity = 0.5
damping = 0.998  # Slight air resistance

# Lists to store the pendulum positions for animation
history_x = []
history_y = []

# Simulate the physics first
current_angle = angle
current_vel = angular_velocity
frames = 200

for _ in range(frames):
    # Physics calculation
    angular_acceleration = (-gravity / length) * math.sin(current_angle)
    current_vel += angular_acceleration
    current_vel *= damping
    current_angle += current_vel

    # Calculate coordinates
    x = origin_x + length * math.sin(current_angle)
    y = origin_y + length * math.cos(current_angle)
    history_x.append(x)
    history_y.append(y)

# Setup the plot for animation
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(0, 600)
ax.set_ylim(600, 0)  # Inverted y-axis to match tkinter's coordinate system
ax.set_aspect('equal')
ax.axis('off')

# Elements to animate
rod, = ax.plot([], [], color='#333333', linewidth=3)
bob, = ax.plot([], [], marker='o', color='#ff4757', markersize=30)

def init():
    rod.set_data([], [])
    bob.set_data([], [])
    return rod, bob

def update(frame):
    x = history_x[frame]
    y = history_y[frame]
    rod.set_data([origin_x, x], [origin_y, y])
    bob.set_data([x], [y])
    return rod, bob

# Create animation
ani = FuncAnimation(fig, update, frames=frames, init_func=init, blit=True, interval=16)
plt.close(fig)

# Display the animation in the notebook
HTML(ani.to_html5_video())
```
