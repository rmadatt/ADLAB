```python
import plotly.express as px
import pandas as pd

# Define the hierarchical AI Tech Stack structure with corresponding languages and tools
data = {
    "Hierarchy": [
        # Silicon & Hardware
        "AI Tech Stack", "Silicon & Hardware", "Silicon & Hardware",
        # Compute Platforms
        "AI Tech Stack", "Compute Platforms & Drivers", "Compute Platforms & Drivers", "Compute Platforms & Drivers",
        # Foundation Models
        "AI Tech Stack", "Foundation Models", "Foundation Models",
        # Model Serving
        "AI Tech Stack", "Model Serving & Repos", "Model Serving & Repos", "Model Serving & Repos",
        # Orchestration
        "AI Tech Stack", "Orchestration & Agents", "Orchestration & Agents",
        # Applications & Vibe Coding
        "AI Tech Stack", "Applications & Vibe Coding", "Applications & Vibe Coding"
    ],
    "Layer": [
        # Silicon & Hardware
        "Silicon & Hardware", "Nvidia / AMD Chips", "Hardware Design (HDLs)",
        # Compute Platforms
        "Compute Platforms & Drivers", "CUDA & ROCm Kernels", "PyTorch & JAX", "Triton Compiler",
        # Foundation Models
        "Foundation Models", "Transformers & LLMs", "Training Scripts",
        # Model Serving
        "Model Serving & Repos", "vLLM / llama.cpp Engines", "Hugging Face Hub", "Serving APIs",
        # Orchestration
        "Orchestration & Agents", "LangChain / LlamaIndex", "Vector DBs & RAG",
        # Applications & Vibe Coding
        "Applications & Vibe Coding", "Vibe Coding Platforms (Cursor/v0)", "Web & App Generation"
    ],
    "Details": [
        # Silicon & Hardware
        "Silicon & Hardware", "GPUs, TPUs, TSMC Wafer Manufacturing", "Verilog, SystemVerilog, VHDL, Chisel",
        # Compute Platforms
        "Compute Platforms & Drivers", "CUDA C++, AMD HIP, C", "PyTorch (Python/C++), JAX", "Triton (Python-like GPU DSL)",
        # Foundation Models
        "Foundation Models", "GPT-4o, Claude, DeepSeek, Llama", "Python (PyTorch, JAX)",
        # Model Serving
        "Model Serving & Repos", "C++, Rust, CUDA", "Python, Rust", "Go, Python, TypeScript",
        # Orchestration
        "Orchestration & Agents", "Python, TypeScript/JavaScript", "Python, C++ (Faiss), Rust",
        # Applications & Vibe Coding
        "Applications & Vibe Coding", "TypeScript, React, Python", "TypeScript, HTML/CSS, Python, Swift"
    ],
    "Weight": [
        # Stack weight visual proportion
        0, 15, 10,
        0, 12, 10, 8,
        0, 18, 12,
        0, 12, 8, 5,
        0, 12, 8,
        0, 15, 15
    ]
}

df = pd.DataFrame(data)

# Generate Treemap
fig = px.treemap(
    df,
    path=["Hierarchy", "Layer", "Details"],
    values="Weight",
    color="Layer",
    title="AI Tech Stack Hierarchy: Tools & Programming Languages",
    color_discrete_sequence=px.colors.qualitative.Pastel
)

# Customize hover info and layout styling
fig.update_traces(root_color="lightgrey")
fig.update_layout(
    margin=dict(t=50, l=10, r=10, b=10),
    font=dict(size=14)
)

fig.show()
```
